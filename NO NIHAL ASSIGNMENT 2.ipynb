```python
#Parent Class
class Car:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def display_info(self):
        print("Brand:", self.brand)
        print("Model:", self.model)


#Child Class
class ElectricCar(Car):
    def __init__(self, brand, model, battery_size):
        super().__init__(brand, model)  # Call parent constructor
        self.battery_size = battery_size  # New attribute

    def display_battery(self):
        print("Battery Size:", self.battery_size, "kWh")



car1 = ElectricCar("Tesla", "Model S", 100)

car1.display_info()      # From Car class
car1.display_battery()   # From ElectricCar class
```

```python
# Parent class
class Computer:
    def __init__(self, brand, processor, ram):
        self.brand = brand
        self.processor = processor
        self.ram = ram

    def display_info(self):
        print(f"Brand: {self.brand}")
        print(f"Processor: {self.processor}")
        print(f"RAM: {self.ram}GB")


# Child class
class Laptop(Computer):
    def __init__(self, brand, processor, ram, battery_life):
        
        super().__init__(brand, processor, ram)
        self.battery_life = battery_life

    def display_info(self):
        
        super().display_info()
        print(f"Battery Life: {self.battery_life} hours")


# Creating an object
laptop1 = Laptop("Dell", "Intel i7", 16, 8)
laptop1.display_info()
```

```python
# Parent class
class ElectronicDevice:
    def __init__(self, brand):
        self.brand = brand

#Child Class 1
class Smartphone(ElectronicDevice):
    def __init__(self, brand, storage, operating_system):
        super().__init__(brand)
        self.storage = storage
        self.operating_system = operating_system

#Child Class 2
class Tablet(ElectronicDevice):
    def __init__(self, brand, screen_size, operating_system):
        super().__init__(brand)
        self.screen_size = screen_size
        self.operating_system = operating_system

#Objects
x = Smartphone("Samsung Galaxy S24", "256GB", "Android 14")
print(x.brand, x.storage, x.operating_system)
```

```python
class Artist:
    def __init__(self, name, age):
        self.name = name
        self.age = age

class Musician(Artist):
    def __init__(self, name, age, instrument_played):
        super().__init__(name, age)
        self.instrument_played = instrument_played

x = Musician("Arijit Singh", 36, "Piano")
print(x.name, x.age, x.instrument_played)

x2 = Musician("Shreya Ghoshal", 39, "Harmonium")
print(x2.name, x2.age, x2.instrument_played)
```

```python
# Parent class
class FictionBook:
    def __init__(self, title, author):
        self.title = title
        self.author = author

# Child class
class MysteryBook(FictionBook):
    def __init__(self, title, author, genre):
        super().__init__(title, author)
        self.genre = genre

# Creating object
b1 = MysteryBook("The Silent Patient", "Alex Michaelides", "Psychological Thriller")
print("title:",b1.title,",","Author:", b1.author,",", "genre:",b1.genre)
```

```python
class Student:
    def __init__(self, name, marks1, marks2, marks3):
        self.name = name
        self.marks1 = marks1
        self.marks2 = marks2
        self.marks3 = marks3

    def print_average(self):
        average = (self.marks1 + self.marks2 + self.marks3) / 3
        print(f"{self.name}'s Average Marks: {average}")

# Creating object
s1 = Student("Priyanshi", 78, 85, 92)
s1.print_average()
```
