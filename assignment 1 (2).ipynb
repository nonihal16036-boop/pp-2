```python
class Laptop:
    def __init__(self, price):
        self.__price = price   # Private attribute

    # Setter method
    def set_price(self, new_price):
        if new_price > 0:
            self.__price = new_price
        else:
            print("Price must be greater than 0")

    # Getter method
    def get_price(self):
        return self.__price


# Example usage
l1 = Laptop(50000)

print("Original Price:", l1.get_price())

l1.set_price(55000)
print("Updated Price:", l1.get_price())
```

```python
class BankAccount:
    def __init__(self, initial_balance=0):
        self.__balance = initial_balance   # Private attribute

    # Deposit method
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited: {amount}")
        else:
            print("Deposit amount must be positive")

    # Withdraw method
    def withdraw(self, amount):
        if amount <= 0:
            print("Withdrawal amount must be positive")
        elif amount > self.__balance:
            print("Insufficient balance")
        else:
            self.__balance -= amount
            print(f"Withdrawn: {amount}")

    # Check balance method
    def get_balance(self):
        return self.__balance


# Example usage
account = BankAccount(1000)

account.deposit(500)
account.withdraw(300)
print("Current Balance:", account.get_balance())
```

```python
class Movie:
    def __init__(self, rating):
        self.__rating = None
        self.set_rating(rating)   # Use setter for validation

    # Setter method
    def set_rating(self, rating):
        if 1 <= rating <= 10:
            self.__rating = rating
        else:
            print("Rating must be between 1 and 10")

    # Getter method
    def get_rating(self):
        return self.__rating


# Example usage
m1 = Movie(8)
print("Movie Rating:", m1.get_rating())

m1.set_rating(11)   # Invalid rating
m1.set_rating(9)    # Valid rating
print("Updated Rating:", m1.get_rating())
```
